```python
def clean_review(text):
    text = re.sub(r'‚Äô', "'", text) # re.sub(pattern, replacement, string)
    text = re.sub(r'-', "", text)    # The re.sub() function in Python is used to replace occurrences of a pattern in a string with a specified replacement
    text = re.sub(r'Äîthey',"they", text)
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'([.!?])([A-Z])', r'\1 \2', text)
    return text.strip()
```

- pattern: The regular expression pattern you want to match.
- repl: The replacement value. This can be a simple string or a function (callback) that receives a match object and returns a string.
- string: The original string where the search and replacement will occur.

```python
clean_review(review_text[0])
```

```python
final_review = []
for i in review_text:
  final_review.append(clean_review(i))
```
